## OS, Systems, kernels, CLIs, and GUIs
- An os is a set of software functions that enable user to manipulate programs and files on their computer 
- **kernel** a low level set of commands that has total control over the computer 
- **drivers** low level programs that interface between kernel and hardware, for example memory, pen, keyboard, mouse
- **cli** - command line interface, or shell. surrounds the kernel, and gives is human readable stuff. 
- usually two ways of doing things on a computer, GUI or CLI
