# Memory paradigms 
## Shared memory  
- shared memeory, all processors share the same address space 
- OpenMP directives best programming 
- all process has access to a pool of shared memeory
- shared memeory is easy to program, but hard to build 
     -bus based sysyems can become saturted 
     -large fats corssbars are expensive 
     - cache-coherency is hard to maintain at scale 


## Distributed memory 
- every processor has its own address space 
- MPI, message passing interface. 
https://en.wikipedia.org/wiki/OpenMP


- MPI is a method of spreading data between machines, which are running the same code 
- you can choose the number of machines, nodes, and compute cores
- each processor has its own local memeory 
- python wrapper for this 
- point to point processsors
- easy to build, bunch of PCs, but hard to program (MPI)
    - you have to spell it all out 
    - higher latency 
    - makes parallel algoroithim development and programming harder
